```python
import numpy as np
```

# Step 1 : Prepare Training Data

```python
# No.of hours  studied 
X=np.array([1,2,3,4])
y=np.array([0,0,1,1])
X,y
```

# Step 2 : Initialize Parameters

```python
w=0.0
b=0.0
learning_rate=1
epochs=5
print("Inital weight :",w)
print("Inital bias : ",b)
```

Define the step or Activation function

```python
def step_function(z):
    return 1 if z>=0 else 0
```

Training loop

```python
for epoch in range(epochs):
    print("Epoch count :",epoch+1)
    for i in range(len(X)):
        # 1. weighted sum 
        z=w*X[i]+b
        # 2. Activation
        y_pred=step_function(z)
        # 3. error
        error=y[i]-y_pred
        # 4. update
        w = w + learning_rate*error*X[i]
        b = b + learning_rate*error
        print(f"Input {X[i]} ,Predicted :{y_pred},Actual : {y[i]}, w: {w},b: {b}")
    

```

Final model parameters

```python
print("Final weights :",w)
print("Final bias :",b)
```

# Testing the model


```python
test_val=int(input("Enter the no of hours studied"))
z=w*test_val+b
prediction=step_function(z)
print("prediction :",prediction)
```

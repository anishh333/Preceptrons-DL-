# XOR gate using ANN

```python
import numpy as np

def step_function(x):
    return 1 if x >= 0 else 0

# 1. Initialize data (X: inputs, y_true: expected labels)
# Example: Solving an XOR-like problem
X = [[0,0], [0,1], [1,0], [1,1]]
y_true = [0, 1, 1, 0]

# 2. Initialize Weights and Biases 
w1,w2,w3,w4,w5,w6=1,1,1,1,1,-2
b1=-0.5
b2=-1.5
b3=-0.5


for i, (x1, x2) in enumerate(X):
    # --- FORWARD PASS ---
    h1 = w1*x1 + w2*x2 + b1
    h2 = w3*x1 + w4*x2 + b2
    ah1 = step_function(h1)
    ah2 = step_function(h2)
        
    y_pred_linear = w5*ah1 + w6*ah2 + b3
    ay = step_function(y_pred_linear)
        

# 4. Final Prediction Check
print("Final Predictions:")
for x1, x2 in X:
    ah1 = step_function(w1*x1 + w2*x2 + b1)
    ah2 = step_function(w3*x1 + w4*x2 + b2)
    final_y = step_function(w5*ah1 + w6*ah2 + b3)
    print(f"Input: [{x1}, {x2}] -> Predicted: {final_y}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Setup the XOR Data
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])

# 2. Define your Weights and Biases
w1, w2, w3, w4, w5, w6 = 1, 1, 1, 1, 1, -2
b1, b2, b3 = -0.5, -1.5, -0.5

# 3. Create a Prediction Function (supporting arrays for the plot)
def predict_xor(x1, x2):
    # Hidden Layer logic
    h1 = (w1 * x1) + (w2 * x2) + b1
    h2 = (w3 * x1) + (w4 * x2) + b2
    
    # Activation (Step Function)
    h1_act = np.where(h1 >= 0, 1, 0)
    h2_act = np.where(h2 >= 0, 1, 0)
    
    # Output Layer logic
    y_out = np.where(((w5 * h1_act) + (w6 * h2_act) + b3) >= 0, 1, 0)
    return y_out

# 4. Create a Grid for the Boundary
x_span = np.linspace(-0.5, 1.5, 400)
y_span = np.linspace(-0.5, 1.5, 400)
xx, yy = np.meshgrid(x_span, y_span)

# 5. Predict over the whole grid
Z = predict_xor(xx, yy)

# 6. Build the Plot
plt.figure(figsize=(8, 6))

# Plot the decision regions
plt.contourf(xx, yy, Z, alpha=0.3, cmap='RdYlBu')
plt.contour(xx, yy, Z, colors='black', linewidths=0.5)

# Plot the 4 XOR input points
for i, label in enumerate(y):
    color = 'red' if label == 0 else 'blue'
    plt.scatter(X[i, 0], X[i, 1], c=color, s=200, edgecolors='k', zorder=3)
    plt.text(X[i, 0]+0.05, X[i, 1]+0.05, f"({X[i,0]},{X[i,1]})", fontsize=12)

# Formatting
plt.title("XOR Decision Boundary with Step Function 🚀")
plt.xlabel("Input x1")
plt.ylabel("Input x2")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
